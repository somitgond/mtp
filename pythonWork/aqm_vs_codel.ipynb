## aqm vs codel
1. unbounded data, ran for 500s
2. codel in default configuration, queueSize = 2084
3. AQM using new qth calculation approach, B=2048

Queue size variation in aqm  


```python
data_codel = pd.read_csv("a_results_LN_codel_zc.csv")
data_codel_10 = data_codel[:10]
data_codel_11 = data_codel[10:]

data_aqm = pd.read_csv("a_results_LN_aqm_zc.csv")
data_aqm_10 = data_aqm[:10]
data_aqm_11 = data_aqm[10:]
```

```python
for c in cols:
    c  = c[0]
    plt.scatter(list(data_codel_11["RTT"]), list(data_codel_11[c]), label="codel (linux reno)")
    plt.scatter(list(data_aqm_11["RTT"]), list(data_aqm_11[c]), label = "aqm (linux reno)")
    plt.xlabel("RTT")
    plt.ylabel(c)
    plt.title(f"{c} vs RTT")
    plt.grid()
    plt.legend()
    plt.show()
```

```python
cols = [["Average Throughput(Mbps)", "std avg throughput"], 
        ["Average Goodput(Mbps)", "std goodput"], 
        ["Link Utilization", "std link utilization"],
        ["Queuing Delay(ms)", "std queuing delay"],
        ["Flow Completion Time(s)", "std flow comp time(s)"],
        ["Packet loss %", "std pkt loss"]]
```

```python
for c in cols:
    plt.errorbar(x=list(data_codel_11["RTT"]), y=list(data_codel_11[c[0]]), yerr=list(data_codel_11[c[1]]), fmt='o', capsize=5, label = "codel")
    plt.errorbar(x=list(data_aqm_11["RTT"]), y=list(data_aqm_11[c[0]]), yerr=list(data_aqm_11[c[1]]), fmt='o', capsize=5, label = "aqm")
    plt.xlabel("RTT")
    plt.ylabel(c[0])
    plt.title(f"{c[0]} vs RTT (New Reno)")
    plt.grid()
    plt.legend()
    plt.show()
```
